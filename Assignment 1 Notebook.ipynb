```python
def calculate_decision_stump(x, w, l):
    x = np.asarray(x, dtype=float).reshape(-1)
    w = np.asarray(w, dtype=float).reshape(-1)
    l = np.asarray(l).reshape(-1)

    if not (len(x) == len(w) == len(l)):
        raise ValueError(
            "x, w and l must have the same number of samples."
        )
    min_error = np.inf
    best_thresh = float(x[0])
    best_dir = 1
    feature_range = np.abs(x.max() - x.min())

    if np.isclose(feature_range, 0.0):
        thresholds = np.array([x.min()])
    else:
        interval = feature_range / 100.0
        thresholds = np.arange(
            x.min() - interval * 2,
            x.max() + interval * 2 + interval,
            interval
        )

    for direction in [1, -1]:
        for threshold in thresholds:
            if direction == 1:
                prediction = np.where(
                    x >= threshold,
                    1,
                    -1
                )
            else:
                prediction = np.where(
                    x < threshold,
                    1,
                    -1
                )
            incorrect = (
                prediction != l
            ).astype(int)
            weighted_error = np.sum(
                w * incorrect
            )

            if weighted_error < min_error:
                min_error = weighted_error
                best_thresh = float(threshold)
                best_dir = int(direction)
    return (
        best_thresh,
        best_dir,
        min_error
    )
```

```python
def calculate_alpha(weighted_error):
    weighted_error = float(weighted_error)
    eps = 1e-12
    weighted_error = np.clip(
        weighted_error,
        eps,
        1.0 - eps
    )
    alpha = 0.5 * np.log(
        (1.0 - weighted_error)
        / weighted_error
    )
    return alpha
```

```python
def classify_dataset_against_weak_classifier( x, thresh, direction):
    x = np.asarray(x, dtype=float).reshape(-1)

    if direction == -1:
        classification = np.where( x < thresh, 1, -1)
    else:
        classification = np.where(x >= thresh, 1, -1)
    return classification.astype(int)
```

```python
def update_weights(weight, alpha, classification, label):
    weight = np.asarray(weight, dtype=float).copy()

    classification = np.asarray(classification).reshape(-1)

    label = np.asarray(label).reshape(-1)

    weight = weight * np.exp(-alpha * label * classification)
    return weight

def normalise_weights(weight):

    weight = np.asarray(weight, dtype=float)
    total = np.sum(weight)
    if total <= 0 or not np.isfinite(total):
        raise ValueError("Weights cannot be normalised.")
    return weight / total
```

```python
def train_adaboost_no_visualisation(data, labels, rounds=30):
    data = np.asarray(data, dtype=float)
    labels = np.asarray(labels).astype(int).reshape(-1)

    n_samples, n_features = data.shape
    model_h = np.zeros((rounds, 3), dtype=float)

    model_alpha = np.zeros(rounds, dtype=float)

    model_err = np.full(rounds, np.inf, dtype=float)

    sample_weight = (np.ones(n_samples)/ n_samples)

    weight_history = [sample_weight.copy()]

    for t in range(rounds):
        for feature in range(n_features):
            threshold, direction, weighted_error = (
                calculate_decision_stump(
                    data[:, feature],
                    sample_weight,
                    labels
                )
            )

            if weighted_error < model_err[t]:

                model_err[t] = weighted_error

                model_h[t] = [
                    threshold,
                    feature,
                    direction
                ]

        model_alpha[t] = calculate_alpha(
            model_err[t]
        )

        weak_prediction = (
            classify_dataset_against_weak_classifier(
                data[:, int(model_h[t, 1])],
                model_h[t, 0],
                model_h[t, 2]
            )
        )

        sample_weight = update_weights(
            sample_weight,
            model_alpha[t],
            weak_prediction,
            labels
        )

        sample_weight = normalise_weights(
            sample_weight
        )

        weight_history.append(
            sample_weight.copy()
        )

    return (
        model_h,
        model_alpha,
        model_err,
        np.array(weight_history)
    )
```

```python
def ensemble_scores(data, model_h, model_alpha, number_of_learners=None):

    data = np.asarray(data, dtype=float)

    if number_of_learners is None:
        number_of_learners = len(model_alpha)

    scores = np.zeros(len(data), dtype=float)

    for t in range(number_of_learners):

        weak_prediction = (
            classify_dataset_against_weak_classifier(
                data[:, int(model_h[t, 1])],
                model_h[t, 0],
                model_h[t, 2]))

        scores += (
            model_alpha[t]
            * weak_prediction)
    return scores


def ensemble_predictions(data, model_h, model_alpha, number_of_learners=None, threshold=0.0):
    scores = ensemble_scores(data, model_h, model_alpha, number_of_learners)
    return np.where(scores >= threshold, 1, -1)
```

```python
# Ensemble scores for margin calculation

def task1_ensemble_scores(data, model_h, model_alpha, number_of_learners=None):
    data = np.asarray(data, dtype=float)
    model_alpha = np.asarray(model_alpha, dtype=float)

    if number_of_learners is None:
        number_of_learners = len(model_alpha)
        
    scores = np.zeros(len(data),dtype=float)

    for t in range(number_of_learners):
        threshold = model_h[t, 0]
        feature = int( model_h[t, 1])
        direction = int(model_h[t, 2])

        feature_values = data[:, feature]

        if direction == -1:
            weak_prediction = np.where(feature_values < threshold, 1, -1)
        else:
            weak_prediction = np.where(feature_values >= threshold, 1, -1)

        scores += (model_alpha[t] * weak_prediction)
    return scores
```

```python
# Signed normalised margins

def task1_signed_normalised_margins(data,labels,model_h,model_alpha,number_of_learners=None):

    labels = np.asarray(labels).reshape(-1)

    model_alpha = np.asarray(model_alpha, dtype=float)

    if number_of_learners is None:
        number_of_learners = len(model_alpha)

    scores = task1_ensemble_scores(data,model_h,model_alpha,number_of_learners)

    denominator = np.sum(np.abs(model_alpha[:number_of_learners]))

    if denominator == 0:
        return np.zeros(len(labels))

    margins = (labels * scores/ denominator)

    return margins
```

```python
from sklearn.tree import DecisionTreeClassifier

# manual adaboost using decision trees
def train_manual_adaboost_with_tree(data,labels,rounds=30,max_depth=1):
    data = np.asarray(data, dtype=float)
    labels = np.asarray(labels).astype(int)

    # Initially every training observation receives the same weight
    sample_weight = np.ones(len(data), dtype=float) / len(data)

    # Store the weak classifiers
    learners = []

    # Store their alpha values
    alphas = []

    # Store their weighted errors
    weighted_errors = []

# Boosting rounds
    for t in range(rounds):
        # Create a shallow decision tree. This is the weak classifier
        learner = DecisionTreeClassifier(max_depth=max_depth, random_state=1000 + t)

        # Train the tree using the current AdaBoost sample weights
        learner.fit(data,labels,sample_weight=sample_weight)

        # Predictions from this weak learner
        weak_prediction = learner.predict(data)

#calculate weighted error
        weighted_error = np.sum(sample_weight * (weak_prediction != labels))
        
        if weighted_error >= 0.5:
            break


# Calculate ALpha
        learner_alpha = calculate_alpha(weighted_error)

#Update sample weights
        sample_weight = update_weights(sample_weight,learner_alpha,weak_prediction,labels)
        sample_weight = normalise_weights(sample_weight)

        # Save weak learner
        learners.append(learner)
        alphas.append(learner_alpha)
        weighted_errors.append(weighted_error)

    return (learners,np.asarray(alphas),np.asarray(weighted_errors))
```

```python
# predict using the tree based adaboost ensemble
def predict_manual_tree_adaboost(data,learners,alphas,threshold=0.0):
    score = np.zeros(len(data), dtype=float)

    # Combine predictions using their AdaBoost alpha weights
    for learner, learner_alpha in zip(learners, alphas):
        score += learner_alpha * learner.predict(data)

    # Convert weighted score into final class prediction
    return np.where(score >= threshold, 1, -1)
```

```python
#Modified adaboost with extra focus
def train_adaboost_with_extra_focus(data,labels,rounds=30,extra_focus=1.0):
    data = np.asarray(data, dtype=float)
    labels = np.asarray(labels).astype(int)

    n_samples, n_features = data.shape

    # Store weak classifiers
    model_h = np.zeros((rounds, 3), dtype=float)

    # Store alpha values
    model_alpha = np.zeros(rounds, dtype=float)

    # Store weighted errors
    model_err = np.full(rounds, np.inf, dtype=float)

    # Initially all examples receive
    # the same weight
    sample_weight = np.ones(n_samples, dtype=float) / n_samples


#Boosting Round
    for t in range(rounds):
        
        # Search all available features
        for feature in range(n_features):
            (threshold,direction,weighted_error) = calculate_decision_stump(data[:, feature],sample_weight,labels)

            # Keep the best weak classifier
            if weighted_error < model_err[t]:
                model_err[t] = weighted_error
                model_h[t] = [threshold,feature,direction]

        # Calculate alpha
        model_alpha[t] = calculate_alpha(model_err[t])

        # Generate predictions from selected weak classifier
        weak_prediction = classify_dataset_against_weak_classifier(data[:, int(model_h[t, 1])],model_h[t, 0],model_h[t, 2])

        # Standard AdaBoost weight update
        sample_weight = update_weights(sample_weight,model_alpha[t],weak_prediction,labels)


#Extra focus modification
        
        # Find currently misclassified examples
        if extra_focus != 1.0:
            difficult_mask = (weak_prediction != labels)

        # Give those difficult observations additional weight
            sample_weight[difficult_mask] *= extra_focus

        # Normalise again so weights sum to 1
        sample_weight = normalise_weights(sample_weight)

    return (model_h,model_alpha,model_err)
```

```python
# Compare different extra focus levels
focus_results = []
for focus_factor in [1.0, 1.25, 1.5]:
    (focus_h,focus_alpha,focus_err) = train_adaboost_with_extra_focus(
        X_task_train,y_task_train,rounds=30,extra_focus=focus_factor)

    # Training prediction
    focus_train_pred = ensemble_predictions(X_task_train,focus_h,focus_alpha)

    # Test prediction
    focus_test_pred = ensemble_predictions(X_task_test,focus_h,focus_alpha)

    # Save results
    focus_results.append({
        'extra_focus_multiplier': focus_factor,
        'train_accuracy': accuracy_score(y_task_train,focus_train_pred),
        'test_accuracy': accuracy_score(y_task_test,focus_test_pred)
    })

# Convert results to table
focus_results_df = pd.DataFrame(focus_results)

print("\nEffect of giving difficult examples extra focus:")
display(focus_results_df)
```

```python
# PREPROCESSING
def task2_make_encoder():
    try:
        return OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        )
    except TypeError:
        return OneHotEncoder(
            handle_unknown="ignore",
            sparse=False
        )


def task2_build_preprocessor(frame, feature_columns):
    numeric_columns = [
        column for column in feature_columns
        if pd.api.types.is_numeric_dtype(frame[column])
    ]

    categorical_columns = [
        column for column in feature_columns
        if column not in numeric_columns
    ]

    numeric_pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="median"))
    ])

    categorical_pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", task2_make_encoder())
    ])

    return ColumnTransformer(
        transformers=[
            ("numeric", numeric_pipeline, numeric_columns),
            ("categorical", categorical_pipeline, categorical_columns)
        ],
        remainder="drop"
    )
```

```python
def task2_calculate_alpha(weighted_error):
    weighted_error = float(weighted_error)
    weighted_error = np.clip(weighted_error, 1e-12, 1 - 1e-12)

    return 0.5 * np.log(
        (1 - weighted_error) / weighted_error
    )


def task2_classify_weak_classifier(x, threshold, direction):
    x = np.asarray(x, dtype=float).reshape(-1)

    if direction == -1:
        prediction = np.where(x < threshold, 1, -1)
    else:
        prediction = np.where(x >= threshold, 1, -1)

    return prediction.astype(int)


def task2_update_weights(weight, alpha, classification, label):
    weight = np.asarray(weight, dtype=float).copy()
    classification = np.asarray(classification, dtype=float).reshape(-1)
    label = np.asarray(label, dtype=float).reshape(-1)

    weight *= np.exp(
        -alpha * label * classification
    )

    return weight


def task2_normalise_weights(weight):
    weight = np.asarray(weight, dtype=float)
    total = np.sum(weight)

    if total <= 0 or not np.isfinite(total):
        raise ValueError("Invalid sample weights.")

    return weight / total


def task2_prepare_stump_cache(data):
    data = np.asarray(data, dtype=float)
    n_samples, n_features = data.shape

    cache = []

    for feature in range(n_features):
        values = data[:, feature]
        order = np.argsort(values)
        sorted_values = values[order]

        change_positions = (
            np.where(np.diff(sorted_values) != 0)[0] + 1
        )

        split_positions = np.concatenate([
            [0],
            change_positions,
            [n_samples]
        ])

        value_range = sorted_values[-1] - sorted_values[0]

        epsilon = max(
            np.finfo(float).eps,
            abs(value_range) * 1e-12
        )

        thresholds = []

        for split in split_positions:
            if split == 0:
                threshold = sorted_values[0] - epsilon

            elif split == n_samples:
                threshold = sorted_values[-1] + epsilon

            else:
                threshold = (
                    sorted_values[split - 1]
                    + sorted_values[split]
                ) / 2

            thresholds.append(threshold)

        cache.append({
            "order": order,
            "split_positions": split_positions,
            "thresholds": np.asarray(thresholds)
        })

    return cache


def task2_calculate_stump(weight, label, feature_cache):
    order = feature_cache["order"]
    split_positions = feature_cache["split_positions"]
    thresholds = feature_cache["thresholds"]

    sorted_weight = np.asarray(weight, dtype=float)[order]
    sorted_label = np.asarray(label).reshape(-1)[order]

    positive_weight = sorted_weight * (sorted_label == 1)
    negative_weight = sorted_weight * (sorted_label == -1)

    positive_prefix = np.concatenate([
        [0.0],
        np.cumsum(positive_weight)
    ])

    negative_prefix = np.concatenate([
        [0.0],
        np.cumsum(negative_weight)
    ])

    positive_left = positive_prefix[split_positions]
    negative_left = negative_prefix[split_positions]

    total_positive = positive_prefix[-1]
    total_negative = negative_prefix[-1]

    error_positive = (
        positive_left
        + total_negative
        - negative_left
    )

    error_negative = (
        negative_left
        + total_positive
        - positive_left
    )

    index_positive = int(np.argmin(error_positive))
    index_negative = int(np.argmin(error_negative))

    best_error_positive = float(
        error_positive[index_positive]
    )

    best_error_negative = float(
        error_negative[index_negative]
    )

    if best_error_positive <= best_error_negative:
        return (
            float(thresholds[index_positive]),
            1,
            best_error_positive
        )

    return (
        float(thresholds[index_negative]),
        -1,
        best_error_negative
    )


def task2_train_adaboost(
    data,
    labels,
    rounds=30,
    learning_rate=1.0
):
    data = np.asarray(data, dtype=float)
    labels = np.asarray(labels, dtype=int).reshape(-1)

    n_samples, n_features = data.shape

    sample_weight = (
        np.ones(n_samples, dtype=float) / n_samples
    )

    stump_cache = task2_prepare_stump_cache(data)

    classifiers = []
    alphas = []
    errors = []

    for round_number in range(int(rounds)):
        best_error = np.inf
        best_feature = None
        best_threshold = None
        best_direction = None

        for feature in range(n_features):
            threshold, direction, weighted_error = (
                task2_calculate_stump(
                    sample_weight,
                    labels,
                    stump_cache[feature]
                )
            )

            if weighted_error < best_error:
                best_error = weighted_error
                best_feature = feature
                best_threshold = threshold
                best_direction = direction

        if best_feature is None or best_error >= 0.5:
            break

        alpha = (
            task2_calculate_alpha(best_error)
            * learning_rate
        )

        weak_prediction = (
            task2_classify_weak_classifier(
                data[:, best_feature],
                best_threshold,
                best_direction
            )
        )

        classifiers.append((
            best_threshold,
            best_feature,
            best_direction
        ))

        alphas.append(alpha)
        errors.append(best_error)

        if best_error <= 1e-12:
            break

        sample_weight = task2_update_weights(
            sample_weight,
            alpha,
            weak_prediction,
            labels
        )

        sample_weight = task2_normalise_weights(
            sample_weight
        )

    return (
        classifiers,
        np.asarray(alphas),
        np.asarray(errors)
    )


def task2_ensemble_scores(data, classifiers, alphas):
    data = np.asarray(data, dtype=float)
    scores = np.zeros(len(data), dtype=float)

    for (
        threshold,
        feature,
        direction
    ), alpha in zip(classifiers, alphas):

        prediction = task2_classify_weak_classifier(
            data[:, feature],
            threshold,
            direction
        )

        scores += alpha * prediction

    return scores


def task2_raw_probability(scores):
    scores = np.asarray(scores, dtype=float)
    value = np.clip(2 * scores, -35, 35)

    return 1 / (
        1 + np.exp(-value)
    )
```

```python
def task2_cross_validate(
    frame,
    features,
    rounds=30,
    learning_rate=1.0,
    folds=3,
    return_oof=False
):
    y_binary = frame[TARGET].astype(int).to_numpy()

    y_boost = np.where(
        y_binary == 1,
        1,
        -1
    )

    splitter = StratifiedKFold(
        n_splits=folds,
        shuffle=True,
        random_state=RANDOM_STATE
    )

    fold_auc = []

    oof_scores = np.full(
        len(frame),
        np.nan
    )

    for fold_number, (
        train_index,
        valid_index
    ) in enumerate(
        splitter.split(frame, y_binary),
        start=1
    ):
        fold_train = frame.iloc[train_index]
        fold_valid = frame.iloc[valid_index]

        preprocessor = task2_build_preprocessor(
            fold_train,
            features
        )

        X_train = preprocessor.fit_transform(
            fold_train[features]
        )

        X_valid = preprocessor.transform(
            fold_valid[features]
        )

        classifiers, alphas, errors = (
            task2_train_adaboost(
                X_train,
                y_boost[train_index],
                rounds=rounds,
                learning_rate=learning_rate
            )
        )

        scores = task2_ensemble_scores(
            X_valid,
            classifiers,
            alphas
        )

        auc = roc_auc_score(
            y_binary[valid_index],
            scores
        )

        fold_auc.append(auc)
        oof_scores[valid_index] = scores

        print(
            f"Fold {fold_number}: "
            f"AUC = {auc:.5f}"
        )

    result = {
        "mean_auc": np.mean(fold_auc),
        "std_auc": np.std(
            fold_auc,
            ddof=1
        ),
        "fold_auc": fold_auc
    }

    if return_oof:
        result["oof_scores"] = oof_scores

    return result
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, roc_auc_score

if 'train_work' not in globals():
    train_work = train_df.copy()
# Target
y = (train_work[TARGET].astype(int))

# Keep existing experiment history
if 'experiment_log' not in globals():
    experiment_log = []


# Create train/validation split if needed
if ('train_indices' not in globals() or 'validation_indices' not in globals()):
    train_indices, validation_indices = train_test_split(
        np.arange(len(train_work)),
        test_size=0.20,
        random_state=RANDOM_STATE,
        stratify=y
    )

# One hot encoder 
def make_one_hot_encoder():
    try:
        return OneHotEncoder(
            handle_unknown='ignore',
            sparse_output=False
        )

    except TypeError:
        return OneHotEncoder(
            handle_unknown='ignore',
            sparse=False
        )

# Build Preprocessor

def build_preprocessor(frame, features):
    # Numeric columns
    numeric_features = [
        column
        for column in features
        if pd.api.types.is_numeric_dtype(
            frame[column]
        )
    ]

    # Categorical columns
    categorical_features = [
        column
        for column in features
        if column not in numeric_features
    ]

    # Numeric preprocessing
    numeric_pipeline = Pipeline([
        (
            'imputer',SimpleImputer(strategy='median')),
        ('scaler',StandardScaler())
    ])

    # Categorical preprocessing
    categorical_pipeline = Pipeline([
        (
            'imputer',SimpleImputer(strategy='most_frequent')
        ),
        (
            'onehot',make_one_hot_encoder()
        )
    ])

    return ColumnTransformer([
        (
            'numeric',
            numeric_pipeline,
            numeric_features
        ),
        (
            'categorical',
            categorical_pipeline,
            categorical_features
        )
    ])
```

```python
# Create a model
def make_model(model_name, params=None):

    if params is None:
        params = {}

    if model_name == 'Naive Bayes':
        return GaussianNB(
            **params
        )

    if model_name == 'kNN':
        return KNeighborsClassifier(
            **params
        )

    if model_name == 'Decision Tree':
        return DecisionTreeClassifier(
            random_state=RANDOM_STATE,
            **params
        )

    if model_name == 'Random Forest':
        return RandomForestClassifier(
            random_state=RANDOM_STATE,
            n_jobs=-1,
            **params
        )

    raise ValueError(
        f'Unknown model: {model_name}'
    )
```

```python
# Reusable experiment function

def evaluate_experiment(
    experiment_name,
    stage,
    model_name,
    params,
    features
):

    # Create model
    model = make_model(
        model_name,
        params
    )

    # Create preprocessing
    preprocessor = build_preprocessor(
        train_work,
        features
    )

    # Create complete pipeline
    pipeline = Pipeline([
        (
            'preprocess',
            preprocessor
        ),
        (
            'model',
            model
        )
    ])

    # Train / validation data
    X_train = (
        train_work
        .iloc[train_indices][features]
    )

    X_valid = (
        train_work
        .iloc[validation_indices][features]
    )

    y_train = y.iloc[
        train_indices
    ]

    y_valid = y.iloc[
        validation_indices
    ]

    # Train model

    pipeline.fit(
        X_train,
        y_train
    )

    # Predictions
    predicted_class = pipeline.predict(
        X_valid
    )

    predicted_probability = (
        pipeline.predict_proba(
            X_valid
        )[:, 1]
    )

    # Metrics

    accuracy = accuracy_score(
        y_valid,
        predicted_class
    )

    auc = roc_auc_score(
        y_valid,
        predicted_probability
    )

    # Record experiment

    result = {
        'experiment': len(experiment_log) + 1,
        'stage': stage,
        'description': experiment_name,
        'model': model_name,
        'parameters': str(params),
        'feature_count': len(features),
        'accuracy': accuracy,
        'roc_auc': auc
    }

    experiment_log.append(
        result
    )

    print(
        f"{result['experiment']:02d}. "
        f"{experiment_name}: "
        f"Accuracy={accuracy:.5f}, "
        f"ROC AUC={auc:.5f}"
    )

    return (
        pipeline,
        result
    )


print("evaluate_experiment() is ready.")
```

```python
def add_engineered_features(frame):
    frame = frame.copy()

    # Loan interest burden 
    if {'loan_amnt', 'loan_int_rate'} <= set(frame.columns):
        frame['loan_interest_burden'] = (frame['loan_amnt']* frame['loan_int_rate']/ 100)

    # Loan-to-income ratio
    if {'loan_amnt', 'person_income'} <= set(frame.columns):
        income = (frame['person_income'].replace(0, np.nan))
        frame['loan_income_ratio_calc'] = (frame['loan_amnt']/ income)

    # Credit history relative to age
    if {'cb_person_cred_hist_length', 'person_age'} <= set(frame.columns):
        age = (frame['person_age'].replace(0, np.nan))
        frame['credit_history_age_ratio'] = (frame['cb_person_cred_hist_length']/ age)

    # Employment length relative to age
    if {'person_emp_length', 'person_age'} <= set(frame.columns):
        age = (frame['person_age'].replace(0, np.nan))
        frame['employment_age_ratio'] = (frame['person_emp_length']/ age)
        
    frame.replace([np.inf, -np.inf],np.nan,inplace=True)
    return frame
```
